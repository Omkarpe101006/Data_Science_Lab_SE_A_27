```python
import numpy as np
import matplotlib.pyplot as plt

# 2. Generate a dataset of at least 1000 data points with a normal distribution
# Using mean=0 and standard deviation=1 (standard normal distribution)
data = np.random.normal(loc=0, scale=1, size=10000)

# 3. Create a histogram of the generated dataset
plt.hist(data, bins=50, edgecolor='black', alpha=0.7)

# 4. Add a title to the histogram
plt.title('Normal Distribution Histogram')

# 5. Label the x-axis and y-axis
plt.xlabel('Value')
plt.ylabel('Frequency')

# Add a grid for better readability
plt.grid(axis='y', alpha=0.75)

# 6. Display the plot
plt.show()
print("A histogram of the normal distribution has been generated and displayed.")
```

```python
mean_data = np.mean(data)
median_data = np.median(data)
variance_data = np.var(data)
std_dev_data = np.std(data)

print(f"Mean of the normal distribution dataset: {mean_data:.4f}")
print(f"Median of the normal distribution dataset: {median_data:.4f}")
print(f"Variance of the normal distribution dataset: {variance_data:.4f}")
print(f"Standard Deviation of the normal distribution dataset: {std_dev_data:.4f}")
```

```python
import numpy as np

# 1. Simulate at least 1000 coin tosses (e.g., 10000)
# Represent heads as 1 and tails as 0
total_tosses = 10000
coin_toss_results = np.random.choice([0, 1], size=total_tosses)

# 2. Count the number of heads (1s) and tails (0s)
num_heads = np.sum(coin_toss_results == 1)
num_tails = np.sum(coin_toss_results == 0)

# 3. Calculate the probability of heads and the probability of tails
probability_heads = num_heads / total_tosses
probability_tails = num_tails / total_tosses

# 4. Print the results
print(f"Total number of tosses: {total_tosses}")
print(f"Number of heads: {num_heads}")
print(f"Number of tails: {num_tails}")
print(f"Probability of heads: {probability_heads:.4f}")
print(f"Probability of tails: {probability_tails:.4f}")
```

```python
import matplotlib.pyplot as plt

# 1. Define labels for the bars
labels = ['Heads', 'Tails']

# 2. Define the counts for the bars
counts = [num_heads, num_tails]

# 3. Create a bar chart
plt.figure(figsize=(6, 4))
plt.bar(labels, counts, color=['skyblue', 'lightcoral'])

# 4. Add a title to the bar chart
plt.title('Coin Toss Simulation Results')

# 5. Label the x-axis and y-axis
plt.xlabel('Outcome')
plt.ylabel('Number of Occurrences')

# Add grid for better readability
plt.grid(axis='y', linestyle='--', alpha=0.7)

# 6. Display the plot
plt.show()
print("A bar chart visualizing the coin toss results has been generated and displayed.")
```
