### 1. WAP to compute factorial of a given number

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)

# Example usage:
number = 5
print(f"The factorial of {number} is {factorial(number)}")

number = 0
print(f"The factorial of {number} is {factorial(number)}")
```

### 2. WAP to display table for a given number

```python
def display_table(number):
    print(f"Multiplication table for {number}:")
    for i in range(1, 11):
        print(f"{number} x {i} = {number * i}")

# Example usage:
display_table(7)
```

### 3. WAP to identify the given number is odd or even

```python
def check_odd_even(number):
    if number % 2 == 0:
        print(f"{number} is an even number.")
    else:
        print(f"{number} is an odd number.")

# Example usage:
check_odd_even(4)
check_odd_even(7)
```

### 4. WAP to identify the entered number is palindrome or not.

```python
def is_palindrome(number):
    # Convert number to string to easily reverse it
    s_number = str(number)
    if s_number == s_number[::-1]:
        print(f"{number} is a palindrome.")
    else:
        print(f"{number} is not a palindrome.")

# Example usage:
is_palindrome(121)
is_palindrome(12321)
is_palindrome(12345)
```

### 5. Develop an Expert System that provides simple decision-making.

An expert system is an AI system that mimics the decision-making ability of a human expert. For simple decision-making, you can use a series of `if-elif-else` statements or a rule-based approach.

Here's a basic example of a rule-based expert system for suggesting what to wear based on the weather.

```python
def weather_expert_system(temperature, weather_condition):
    recommendation = ""
    if temperature > 25:
        recommendation = "It's hot! Wear light clothing (shorts, t-shirt)."
    elif temperature >= 15 and temperature <= 25:
        recommendation = "It's moderate. A light jacket or long sleeves would be good."
    else:
        recommendation = "It's cold! Wear warm layers (coat, scarf, gloves)."

    if "rain" in weather_condition.lower():
        recommendation += " Don't forget an umbrella or a raincoat!"
    elif "snow" in weather_condition.lower():
        recommendation += " Make sure to wear snow boots and a heavy coat!"
    elif "sun" in weather_condition.lower() and temperature > 20:
        recommendation += " And don't forget sunscreen!"

    return recommendation

# Example usage:
print(weather_expert_system(28, "sunny"))
print(weather_expert_system(18, "cloudy"))
print(weather_expert_system(5, "rainy"))
print(weather_expert_system(-2, "snowy"))
```
