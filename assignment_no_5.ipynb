```python
import pandas as pd
import numpy as np

print("Libraries pandas and numpy imported successfully.")
```

```python
iris_url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
column_names = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']
df = pd.read_csv(iris_url, names=column_names)
print("Iris dataset loaded successfully into DataFrame 'df'.")
```

```python
print("First 5 rows of the dataset:")
print(df.head())

print("\nInformation about the dataset:")
df.info()

print("\nShape of the dataset (rows, columns):")
print(df.shape)
```

```python
df_filtered_sepal_length = df[df['sepal_length'] > 5]

print("First 5 rows of the dataset filtered by 'sepal_length' > 5:")
print(df_filtered_sepal_length.head())
```

```python
df_filtered_species = df[df['species'] == 'Iris-setosa']

print("First 5 rows of the dataset filtered by 'species' == 'Iris-setosa':")
print(df_filtered_species.head())
```

```python
grouped_by_species = df.groupby('species')
print("DataFrame 'df' grouped by 'species' successfully.")
```

```python
numerical_columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

print("\nMean of numerical columns by species:")
print(grouped_by_species[numerical_columns].mean())

print("\nSum of numerical columns by species:")
print(grouped_by_species[numerical_columns].sum())

print("\nCount of numerical columns by species:")
print(grouped_by_species[numerical_columns].count())

print("\nMinimum of numerical columns by species:")
print(grouped_by_species[numerical_columns].min())

print("\nMaximum of numerical columns by species:")
print(grouped_by_species[numerical_columns].max())
```
